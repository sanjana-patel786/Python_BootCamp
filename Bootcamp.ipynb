```python
#bonus calcualtor

fnum=int(input("enter first number:"))
snum=int(input("enter second number:"))
op=input("enter operation:")
if op=="+":
  print(fnum+snum)
elif op=="-":
  print(fnum-snum)
elif op=="*":
  print(fnum*snum)
elif op=="/":
  print(fnum/snum)
else:
  print("invalid operation")
```
